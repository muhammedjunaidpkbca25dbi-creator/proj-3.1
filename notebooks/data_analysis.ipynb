```python
q("""
  SELECT
       batter AS player,
       SUM(CASE WHEN is_legal = 1 THEN 1 ELSE 0 END)AS total_balls_faced,
       ROUND(100.0*SUM(batter_runs))/SUM(
           CASE
             WHEN is_wide_ball=0 AND is_no_ball=0
             THEN 1
             ELSE 0
           END) AS strike_rate,
       SUM(batter_runs) AS total_runs
  FROM v_ball
  GROUP BY batter
  ORDER BY total_balls_faced DESC ;""")
```

```python
q("""
-- Find an ideal middle-order batter who can lift the run rate in overs 7 to 15. 

SELECT
    batter,
    SUM(batter_runs) AS runs,
    SUM(is_legal) AS balls,
    ROUND(
        100.0 * SUM(batter_runs) / NULLIF(SUM(is_legal), 0),
        2
    ) AS strike_rate
FROM v_ball
WHERE over_number BETWEEN 7 AND 15
GROUP BY batter
HAVING SUM(is_legal) >= 50
ORDER BY strike_rate DESC, runs DESC;
""")
```

```python
q("""
  --Find a finisher : highest strike rate in the last five overs.
  
  SELECT batter,SUM(batter_runs) AS runs,SUM(is_legal) AS balls,
                ROUND (100.0*SUM(batter_runs)/SUM(is_legal),2) AS strike_rate
                
    FROM v_ball
    WHERE phase="Death"
    GROUP BY batter
    HAVING balls>500
    ORDER BY strike_rate DESC
    LIMIT 10;""")
```

```python
# who is the most reliable in a chase - strike rate batting second?

q("""
  SELECT batter,SUM(batter_runs) AS runs,SUM(is_legal) AS balls,
                  ROUND (100.0*SUM(batter_runs)/SUM(is_legal),2) AS strike_rate
                  
      FROM v_ball
      WHERE innings=2
      GROUP BY batter
      HAVING balls>500
      ORDER BY strike_rate DESC
      LIMIT 10;""")
```

```python
# which batter are hardest to bowl a dot ball to in the death overs?

q("""
   SELECT batter,COUNT(*) AS total_balls,
   SUM(CASE WHEN batter_runs = 0 THEN 1 ELSE 0 END) AS dot_balls,
   ROUND(
        100.0*SUM(CASE WHEN batter_runs = 0 THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS dot_ball_percentage
FROM v_ball
WHERE phase = 'Death'
  AND is_legal = 1
GROUP BY batter
HAVING total_balls > 500
ORDER BY dot_ball_percentage ASC
LIMIT 10;
""")
```

```python
# who is the most economical bowler, counting only runs that were their fault.

q("""
  SELECT bowler,
  SUM(is_legal) AS balls,
  ROUND(6.0*SUM(bowler_runs)/SUM(is_legal),2) AS economy_rate
  FROM v_ball
  GROUP BY bowler
  HAVING balls>500
  ORDER BY economy_rate ASC
  LIMIT 10;""")
```

```python
# which bowler takes a wicket most often - best strike rate in balls per wicket?

q("""
  SELECT
    bowler,
    SUM(is_legal) AS total_balls,
    SUM(bowler_wicket) AS total_wickets,
    ROUND(
        1.0 * SUM(is_legal) / SUM(bowler_wicket),
        2
    ) AS balls_per_wicket
FROM v_ball
GROUP BY bowler
HAVING total_wickets > 50
ORDER BY balls_per_wicket ASC
LIMIT 10;
""")
```

```python
# which bowler gives away the fewest boundaries?
q("""
  SELECT
    bowler,
    SUM(is_legal) AS total_balls,
    SUM(CASE WHEN batter_runs IN(4,6) THEN 1 ELSE 0 END) boundaries,
    ROUND(
        100.0 * SUM(CASE WHEN batter_runs IN(4,6) THEN 1 ELSE 0 END)
        /SUM(is_legal),
        2) AS boundary_perc
    FROM v_ball
    WHERE is_legal = 1
    GROUP BY bowler
    HAVING total_balls>1000
    ORDER BY boundary_perc ASC
    LIMIT 10;""")
```

```python
q("""
-- Which spinner is the best wicket-taker?

SELECT
    bowler,
    SUM(bowler_wicket) AS wicket,
    ROUND(
        100.0 * SUM(bowler_wicket) / SUM(is_legal),
        2
    ) AS wicket_perc
FROM v_ball
WHERE bowler_type LIKE '%Legbreak%'
   OR bowler_type LIKE '%Offbreak%'
   OR bowler_type LIKE '%Orthodox%'
   OR bowler_type LIKE '%Wrist spin%'
   OR bowler_type LIKE '%Googly%'
GROUP BY bowler
HAVING SUM(is_legal) > 500
ORDER BY wicket DESC
LIMIT 10;""")
```

```python
# find a genuine all-rounder : 1000+ runs and 50+ wickets

q("""
SELECT
    b.batter AS player,
    SUM(b.batter_runs) AS runs,
    (SELECT SUM(x.bowler_wicket) FROM v_ball x
    WHERE x.bowler=b.batter) AS wickets
FROM v_ball b
GROUP BY b.batter
HAVING runs>=1000 AND wickets>=50
ORDER BY runs DESC
LIMIT 10;""") 
```

```python
# who performs best when chasing at the death?

q("""
SELECT 
    batter,
    SUM(is_legal) AS balls,
        ROUND(100.0*SUM(batter_runs)/SUM(is_legal),2) AS strike_rate,
        SUM(batter_runs) AS runs
FROM v_ball
WHERE phase='Death' AND innings=2
GROUP BY batter
HAVING balls>200
ORDER BY strike_rate DESC
LIMIT 10;""")
```
